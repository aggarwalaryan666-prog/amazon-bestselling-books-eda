```python
import pandas as pd
df=pd.read_csv('books.csv')
df.head()
df.columns
print("Average Rating by Genre:")
print(df.groupby('Genre')['User Rating'].mean().round(2))

print("\nAverage Price by Genre:")
print(df.groupby('Genre')['Price'].mean().round(2))

print("\nAverage Reviews by Genre:")
print(df.groupby('Genre')['Reviews'].mean().round(2))

print("\nMost Reviewed Book:")
print(
    df.loc[
        df['Reviews'].idxmax(),
        ['Name', 'Author', 'Reviews', 'User Rating']
    ]
)

print("\nHighest Rated Books:")
print(
    df[df['User Rating'] == df['User Rating'].max()][
        ['Name', 'Author', 'User Rating']
    ].head(10)
)

print("\nCorrelation between Reviews and Rating:")
print(
    round(
        df['Reviews'].corr(df['User Rating']),
        2
    )
)

print("\nCorrelation between Price and Rating:")
print(
    round(
        df['Price'].corr(df['User Rating']),
        2
    )
)
```

## Key Findings

- Fiction books have a slightly higher average user rating (4.65) compared to Non Fiction books (4.60).
- Non Fiction books have a higher average price (₹14.84) compared to Fiction books (₹10.85).
- Fiction books receive more average reviews than Non Fiction books in this dataset.
- "Where the Crawdads Sing" by Delia Owens is the most reviewed book, with 87,841 reviews.
- The correlation between book price and user rating is approximately 0, indicating that price does not have a strong linear relationship with user ratings.
- Overall, user ratings remain consistently high across both genres, suggesting that genre and price alone are not strong predictors of book ratings.

## Overall Conclusion

The analysis shows that Amazon bestselling books generally maintain high user ratings across both Fiction and Non Fiction categories. Non Fiction books tend to be more expensive, while Fiction books receive more reviews on average. However, the weak correlation between price and rating suggests that a higher book price does not necessarily result in a higher user rating.

This analysis demonstrates how exploratory data analysis can be used to identify patterns and relationships in book sales and customer rating data.
